```python
import requests
```

```python
url = "http://export.arxiv.org/api/query?search_query=cat:cs.CL&start=0&max_results=10"

cevap = requests.get(url)
```

```python
cevap
```

```python
print(cevap.status_code)
```

```python
print(cevap.text[:1000])
```

xml parse

```python
import xml.etree.ElementTree as ET
```

```python
root = ET.fromstring(cevap.text)
```

```python
root
```

```python
print(root.tag)
```

```python
print(len(root))
```

```python
entries = root.findall("entry")
```

```python
entries
```

```python
namespace = {
    "atom": "http://www.w3.org/2005/Atom"
}
entries = root.findall("atom:entry", namespace)
```

```python
print(len(entries))
```
