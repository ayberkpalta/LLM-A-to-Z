```python
import requests
import xml.etree.ElementTree as ET
```

```python
url = "http://export.arxiv.org/api/query?search_query=cat:cs.CL&start=0&max_results=10"

response = requests.get(url)

print(response.status_code)
```

```python
root = ET.fromstring(response.text)

namespace = {
    "atom": "http://www.w3.org/2005/Atom"
}

entries = root.findall("atom:entry", namespace)

print(len(entries))
```

```python
papers = []

for entry in entries:

    title = entry.find("atom:title", namespace)
    abstract = entry.find("atom:summary", namespace)
    paper_id = entry.find("atom:id", namespace)

    paper = {
        "id": paper_id.text,
        "title": title.text,
        "abstract": abstract.text
    }

    papers.append(paper)
```

```python
papers = []

for entry in entries:

    paper_id = entry.find("atom:id", namespace)
    title = entry.find("atom:title", namespace)
    abstract = entry.find("atom:summary", namespace)
    published = entry.find("atom:published", namespace)

    authors = entry.findall("atom:author", namespace)
    author_names = []

    for author in authors:
        name = author.find("atom:name", namespace)
        author_names.append(name.text)

    categories = entry.findall("atom:category", namespace)
    category_names = []

    for category in categories:
        category_names.append(category.get("term"))

    links = entry.findall("atom:link", namespace)

    paper_url = None
    pdf_url = None

    for link in links:
        if link.get("type") == "text/html":
            paper_url = link.get("href")

        elif link.get("type") == "application/pdf":
            pdf_url = link.get("href")

    paper = {
        "id": paper_id.text,
        "title": title.text,
        "abstract": abstract.text,
        "authors": author_names,
        "categories": category_names,
        "published": published.text,
        "paper_url": paper_url,
        "pdf_url": pdf_url
    }

    papers.append(paper)
```

data validation

```python
print("Number of papers:", len(papers))
```

```python
from enum import unique
ids = [paper["id"] for paper in papers]

print("Total IDs:", len(ids))
print("Unique IDs:", len(set(ids)))#tekrar eden ID'leri tekilleştiriyor.
```

```python
required_fields = [
    "id",
    "title",
    "abstract",
    "authors",
    "categories",
    "published",
    "paper_url",
    "pdf_url"
]

for paper in papers:
    for field in required_fields:
        if field not in paper or paper[field] is None:
            print("Missing:", field, "in", paper["id"])
```

```python
for paper in papers:
    for field in required_fields:
        value = paper.get(field)

        if value is None:
            print("None:", field, "->", paper["id"])

        elif isinstance(value, str) and value.strip() == "":
            print("Empty string:", field, "->", paper["id"])

        elif isinstance(value, list) and len(value) == 0:
            print("Empty list:", field, "->", paper["id"])
```

```python
titles = [paper["title"].strip().lower() for paper in papers]

print("Total titles:", len(titles))
print("Unique titles:", len(set(titles)))
```

```python
from collections import Counter

title_counts = Counter(titles)

for title, count in title_counts.items():
    if count > 1:
        print("Duplicate title:", title, "->", count)
```

```python
print("=== DATA VALIDATION REPORT ===")
print("Total papers:", len(papers))
print("Unique IDs:", len(set(ids)))
print("Unique titles:", len(set(titles)))
```

```python
text = papers[0]["abstract"]

from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)

print(tokens)
```

```python
from collections import Counter

token_counts = Counter(tokens)

print(token_counts.most_common(20))
```

```python
import string

tokens_no_punct = [
    word for word in tokens
    if word not in string.punctuation
]

print(tokens_no_punct[:30])
```

```python
from nltk.corpus import stopwords

stopwords_english = set(stopwords.words("english"))
```

```python
tokens_clean = [
    word for word in tokens_no_punct
    if word.lower() not in stopwords_english
]

print(tokens_clean[:30])
```

```python
print(len(tokens))
print(len(tokens_clean))
```

prep function

```python
def preprocess_text(text):
    tokens = word_tokenize(text)

    tokens_no_punct = [
        word for word in tokens
        if word not in string.punctuation
    ]

    tokens_clean = [
        word for word in tokens_no_punct
        if word.lower() not in stopwords_english
    ]

    return tokens_clean
```

```python
cleaned = preprocess_text(papers[0]["abstract"])

print(cleaned[:30])
print("Number of tokens:", len(cleaned))
```

```python
for paper in papers:
    paper["processed_abstract"] = preprocess_text(paper["abstract"])
```

TF-IDF
TF — Term Frequency

Bu kelime bu dokümanda ne kadar geçiyor?

IDF — Inverse Document Frequency

Bu kelime bütün dokümanlarda ne kadar yaygın?

Bu ikisi birleşerek:

TFIDF=TF×IDF

oluşturuyor.

Yani örneğin the gibi kelimeler:

```python
from sklearn.feature_extraction.text import TfidfVectorizer

processed_texts = []

for paper in papers:
    text = " ".join(paper["processed_abstract"])
    processed_texts.append(text)

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(processed_texts)

print(X.shape) # 10 makale 540 farklı kelime 
```

```python
feature_names = vectorizer.get_feature_names_out()

print(feature_names[:20])
print("Number of features:", len(feature_names))
```

```python
print(X[0].toarray()) # İlk makalenin TF-IDF değerlerini görelim:
```

```python
first_vector = X[0].toarray()[0]

for word, score in zip(feature_names, first_vector):
    if score > 0:
        print(f"{word}: {score:.4f}")
```
