parse entry

```python
def parse_entry(entry, label):
    paper_id = entry.find("atom:id", namespace)
    title = entry.find("atom:title", namespace)
    abstract = entry.find("atom:summary", namespace)
    published = entry.find("atom:published", namespace)

    authors = entry.findall("atom:author", namespace)
    author_names = []

    for author in authors:
        name = author.find("atom:name", namespace)
        if name is not None:
            author_names.append(name.text)

    categories = entry.findall("atom:category", namespace)
    category_names = []

    for category in categories:
        category_names.append(category.get("term"))

    links = entry.findall("atom:link", namespace)

    paper_url = None
    pdf_url = None

    for link in links:
        if link.get("type") == "text/html":
            paper_url = link.get("href")
        elif link.get("type") == "application/pdf":
            pdf_url = link.get("href")

    paper = {
        "id": paper_id.text,
        "title": title.text,
        "abstract": abstract.text,
        "authors": author_names,
        "categories": category_names,
        "published": published.text,
        "paper_url": paper_url,
        "pdf_url": pdf_url,
        "label": label
    }

    return paper
```

1. Exact duplicate
   Aynı ID + aynı label + aynı title

2. Multi-label
   Aynı ID + farklı label

3. Data leakage
   Aynı ID'nin train ve validation/test'te bulunması

```python
# Duplicate ID'leri iki gruba ayıracağız:
# 1. Exact duplicate
# 2. Multi-label

exact_duplicates = []
multi_label = []


for paper_id, count in duplicate_ids.items():

    # Aynı ID'ye sahip paperları bul
    papers = [
        paper
        for paper in all_papers
        if paper["id"] == paper_id
    ]

    # Bu ID'nin sahip olduğu label'ları al
    labels = [
        paper["label"]
        for paper in papers
    ]

    # Tekrarsız label sayısı
    unique_labels = set(labels)

    if len(unique_labels) == 1:
        exact_duplicates.append(paper_id)

    else:
        multi_label.append(paper_id)


print("Exact duplicates:", len(exact_duplicates))
print("Multi-label papers:", len(multi_label))
```

```python
from collections import Counter


# ============================================================
# MULTI-LABEL PAIR ANALYSIS
# ============================================================

label_pairs = Counter()


for paper_id in multi_label:

    papers = [
        paper
        for paper in all_papers
        if paper["id"] == paper_id
    ]

    labels = sorted({
        paper["label"]
        for paper in papers
    })

    # Label çiftini say
    if len(labels) == 2:
        pair = (labels[0], labels[1])
        label_pairs[pair] += 1


print("\n" + "=" * 60)
print("MULTI-LABEL PAIR ANALYSIS")
print("=" * 60)


for pair, count in label_pairs.items():

    print(
        f"{pair[0]} + {pair[1]} : {count}"
    )
```

```python
# ============================================================
# 3+ LABEL'A SAHİP PAPER'LARI BUL
# ============================================================

print("\n" + "=" * 60)
print("PAPERS WITH 3 OR MORE LABELS")
print("=" * 60)

for paper_id in multi_label:

    papers = [
        paper
        for paper in all_papers
        if paper["id"] == paper_id
    ]

    labels = sorted({
        paper["label"]
        for paper in papers
    })

    if len(labels) >= 3:

        print(f"\nID: {paper_id}")
        print(f"Label count: {len(labels)}")
        print(f"Labels: {labels}")

        for paper in papers:
            print("  -", paper["label"], "|", paper["title"])
```

```python
# ============================================================
# LABEL DISTRIBUTION
# ============================================================

label_counts = Counter(
    paper["label"]
    for paper in all_papers
)

print("\n" + "=" * 60)
print("LABEL DISTRIBUTION")
print("=" * 60)

for label, count in label_counts.most_common():

    print(f"{label}: {count}")
```

```python
# ============================================================
# UNIQUE PAPER COUNT PER LABEL
# ============================================================
from collections import Counter, defaultdict

label_unique_ids = defaultdict(set)

for paper in all_papers:

    label = paper["label"]
    paper_id = paper["id"]

    label_unique_ids[label].add(paper_id)


print("\n" + "=" * 60)
print("UNIQUE PAPER COUNT PER LABEL")
print("=" * 60)

for label, paper_ids in label_unique_ids.items():

    print(
        f"{label}: {len(paper_ids)} unique papers"
    )
```

check null value

```python
# ============================================================
# MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUE ANALYSIS")
print("=" * 60)

fields = [
    "id",
    "title",
    "abstract",
    "authors",
    "categories",
    "published",
    "paper_url",
    "pdf_url",
    "label"
]

for field in fields:

    missing_count = sum(
        1
        for paper in all_papers
        if paper.get(field) is None
        or paper.get(field) == ""
    )

    print(f"{field:15} : {missing_count}")
```

```python
# ============================================================
# ABSTRACT LENGTH ANALYSIS
# ============================================================

abstract_lengths = []

for paper in all_papers:

    abstract = paper.get("abstract", "")

    word_count = len(
        abstract.split()
    )

    abstract_lengths.append(word_count)


print("\n" + "=" * 60)
print("ABSTRACT LENGTH ANALYSIS")
print("=" * 60)

print("Minimum words :", min(abstract_lengths))
print("Maximum words :", max(abstract_lengths))
print("Average words :", sum(abstract_lengths) / len(abstract_lengths));
```

```python
# ============================================================
# VERY SHORT ABSTRACTS
# ============================================================

print("\n" + "=" * 60)
print("VERY SHORT ABSTRACTS")
print("=" * 60)

short_abstract_count = 0

for paper in all_papers:

    abstract = paper.get("abstract", "")

    word_count = len(abstract.split())

    if word_count < 50:

        short_abstract_count += 1

        print("\nID:", paper["id"])
        print("Label:", paper["label"])
        print("Word count:", word_count)
        print("Title:", paper["title"])
        print("Abstract:", abstract)


print("\nTotal short abstracts:", short_abstract_count)
```

```python
# ============================================================
# LABEL CONSISTENCY CHECK
# ============================================================

expected_labels = {
    "NLP",
    "Computer Vision",
    "Machine Learning",
    "Robotics"
}

actual_labels = {
    paper["label"]
    for paper in all_papers
}

print("\n" + "=" * 60)
print("LABEL CONSISTENCY")
print("=" * 60)

print("Expected labels:")
print(expected_labels)

print("\nActual labels:")
print(actual_labels)

print("\nUnexpected labels:")
print(actual_labels - expected_labels)

print("\nMissing expected labels:")
print(expected_labels - actual_labels)
```

```python
# ============================================================
# CATEGORIES FORMAT CHECK
# ============================================================

invalid_categories = []

for paper in all_papers:

    categories = paper.get("categories")

    if not isinstance(categories, list):
        invalid_categories.append(paper["id"])


print("\n" + "=" * 60)
print("CATEGORIES FORMAT CHECK")
print("=" * 60)

print("Invalid category fields:", len(invalid_categories))

if invalid_categories:

    for paper_id in invalid_categories:
        print(paper_id)
```

```python
# ============================================================
# MULTI-LABEL PAPERS
# ============================================================

from collections import defaultdict

paper_labels = defaultdict(set)

for paper in all_papers:
    paper_labels[paper["id"]].add(paper["label"])


multi_label_papers = {
    paper_id: labels
    for paper_id, labels in paper_labels.items()
    if len(labels) > 1
}


print("\n" + "=" * 60)
print("MULTI-LABEL PAPERS")
print("=" * 60)

print("Total multi-label papers:", len(multi_label_papers))

for paper_id, labels in multi_label_papers.items():

    print("\nID:", paper_id)
    print("Labels:", sorted(labels))

    for paper in all_papers:

        if paper["id"] == paper_id:
            print("Title:", paper["title"])
            break
```

```python
# ============================================================
# MERGE MULTI-LABEL PAPERS
# ONE PAPER = ONE RECORD
# ============================================================

from collections import defaultdict

merged_papers = {}

for paper in all_papers:

    paper_id = paper["id"]

    # Paper daha önce yoksa temel bilgilerini oluştur
    if paper_id not in merged_papers:

        merged_papers[paper_id] = {
            "id": paper["id"],
            "title": paper["title"],
            "abstract": paper["abstract"],
            "authors": paper["authors"],
            "categories": paper["categories"],
            "published": paper["published"],
            "paper_url": paper["paper_url"],
            "pdf_url": paper["pdf_url"],
            "labels": []
        }

    # Label'ı ekle
    if paper["label"] not in merged_papers[paper_id]["labels"]:

        merged_papers[paper_id]["labels"].append(
            paper["label"]
        )


merged_papers = list(merged_papers.values())
```

```python
# ============================================================
# CHECK MERGED DATASET
# ============================================================

print("\n" + "=" * 60)
print("MERGED DATASET")
print("=" * 60)

print("Original records :", len(all_papers))
print("Unique papers    :", len(merged_papers))
```

```python
# ============================================================
# DUPLICATE RECORD ANALYSIS
# ============================================================

from collections import Counter

id_counts = Counter(
    paper["id"]
    for paper in all_papers
)

duplicate_ids = {
    paper_id: count
    for paper_id, count in id_counts.items()
    if count > 1
}

print("\n" + "=" * 60)
print("DUPLICATE RECORD ANALYSIS")
print("=" * 60)

print("Total records       :", len(all_papers))
print("Unique IDs          :", len(id_counts))
print("Duplicate ID count  :", len(duplicate_ids))

# Kaç tane fazla kayıt oluşmuş?
extra_records = sum(
    count - 1
    for count in duplicate_ids.values()
)

print("Extra duplicate records:", extra_records)
```

```python
# ============================================================
# DUPLICATE FREQUENCY
# ============================================================

print("\n" + "=" * 60)
print("DUPLICATE FREQUENCY")
print("=" * 60)

for paper_id, count in duplicate_ids.items():

    print(f"{count}x | {paper_id}")
```
