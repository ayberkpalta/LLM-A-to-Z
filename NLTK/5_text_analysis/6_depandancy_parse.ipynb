```python
import pandas as pd
df = pd.read_csv('training.1600000.processed.noemoticon.csv.zip',
                 encoding='latin-1',
                 names=['target', 'ids', 'date', 'flag', 'user', 'text'])

df = df[['text']]  
print(df.head())
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

for i, row in df.head(5).iterrows():
    doc = nlp(row['text'])
    print(f"\nTweet: {row['text']}")
    for token in doc:
        print(f"{token.text:<12} | Head: {token.head.text:<12} | Dep: {token.dep_:<12} | POS: {token.pos_}")
```

Step 4: Extract and Store Dependencies


```python
parsed_rows = []

for i, row in df.head(5).iterrows():
    doc = nlp(row['text'])
    deps = [(token.text, token.dep_, token.head.text) for token in doc]
    parsed_rows.append({'text': row['text'], 'dependencies': deps})

parsed_df = pd.DataFrame(parsed_rows)
print(parsed_df.head())
```
