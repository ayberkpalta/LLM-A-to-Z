```python
import nltk
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = "Hello! Welcome to NLP, using NLTK."
```

```python
tokens = word_tokenize(text)

clean_text = [
    word for word in tokens
    if word.isalnum()
]

print(' '.join(clean_text))
```
